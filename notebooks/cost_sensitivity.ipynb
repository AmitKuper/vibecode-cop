# Token Cost and Sensitivity Analysis

`TOTAL_TOKENS` below is the project-wide aggregate token estimate (2,850,905) that
`docs/cost.md` reconciles against — see that file for the per-phase breakdown, the
estimation methodology (session-based estimates; no per-call metering), and the
runtime optimization strategies (template hints, local Ollama, deterministic search
engine = ~zero in-game tokens). This notebook does not assume a provider price;
edit the scenario rates to match the applicable service.


```python
TOTAL_TOKENS = 2_850_905
milestone_weights = {
    "M0 audit": 0.08,
    "M1 composition": 0.30,
    "M2 zero trust": 0.20,
    "M3 RL": 0.18,
    "M4 adaptive MCP": 0.10,
    "M5 verification": 0.10,
    "M6 evidence": 0.04,
}
assert abs(sum(milestone_weights.values()) - 1.0) < 1e-12
{name: round(TOTAL_TOKENS * weight) for name, weight in milestone_weights.items()}
```

## Price sensitivity

The scenarios below are illustrative blended dollars per million tokens, not claims about current pricing.

```python
scenario_usd_per_million = [1, 5, 10, 25, 50]
sensitivity = [
    {
        "usd_per_million_tokens": rate,
        "estimated_total_usd": round(TOTAL_TOKENS / 1_000_000 * rate, 2),
    }
    for rate in scenario_usd_per_million
]
sensitivity
```

## Interpretation

Cost scales linearly with both total tokens and the chosen blended rate. 
For separate input/output pricing, replace the aggregate with actual provider usage fields; 
do not infer a split from this aggregate.

Per-phase allocation of this aggregate and the in-game cost story live in
`docs/cost.md`. The milestone weights above are the design-time plan weights;
`docs/cost.md` carries the by-phase view of how development actually unfolded.
